```python

import nibabel as nib
from pathlib import Path
import numpy as np

patient_dir = Path(r"D:\PS07_CT2MRI\dataset\volumes\P1046")

for filename in ["ct.nii.gz", "mr.nii.gz", "mask.nii.gz"]:

    path = patient_dir / filename
    image = nib.load(str(path))
    data = image.get_fdata()

    print("\n" + "=" * 40)
    print("File:", filename)
    print("Shape:", data.shape)
    print("Data type:", image.get_data_dtype())
    print("Voxel spacing:", image.header.get_zooms()[:3])
    print("Orientation:", nib.aff2axcodes(image.affine))
    print("Value range:", np.min(data), "to", np.max(data))
    print("Affine matrix:\n", image.affine)
```
