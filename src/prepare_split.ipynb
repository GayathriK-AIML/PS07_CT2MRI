```python

import pandas as pd
from pathlib import Path

ROOT = Path(r"D:\PS07_CT2MRI")
DATASET = ROOT / "dataset"
VOLUMES = DATASET / "volumes"

manifest = pd.read_csv(DATASET / "manifest.csv")

train_df = manifest[manifest["split"] == "train"].copy()
val_df = manifest[manifest["split"] == "val"].copy()

# Verify that every manifest patient has a volume folder
missing = [
    pid for pid in manifest["id"]
    if not (VOLUMES / pid).is_dir()
]

print("Training patients:", len(train_df))
print("Validation patients:", len(val_df))
print("Missing volume folders:", missing)

# Save patient lists
train_df.to_csv(ROOT / "train_patients.csv", index=False)
val_df.to_csv(ROOT / "val_patients.csv", index=False)

print("\nSaved:")
print("train_patients.csv")
print("val_patients.csv")

print("\nValidation IDs:")
print(val_df["id"].tolist())
```
