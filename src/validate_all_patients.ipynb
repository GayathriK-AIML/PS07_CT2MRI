```python

import nibabel as nib
import numpy as np
from pathlib import Path

BASE = Path(r"D:\PS07_CT2MRI\dataset\volumes")

patients = sorted(
    p for p in BASE.iterdir() if p.is_dir()
)

valid = []
issues = []

for folder in patients:
    pid = folder.name

    try:
        ct = nib.load(str(folder / "ct.nii.gz"))
        mr = nib.load(str(folder / "mr.nii.gz"))
        mask = nib.load(str(folder / "mask.nii.gz"))

        same_shape = (
            ct.shape == mr.shape == mask.shape
        )

        same_affine = (
            np.allclose(ct.affine, mr.affine)
            and np.allclose(ct.affine, mask.affine)
        )

        if same_shape and same_affine:
            valid.append(pid)
        else:
            issues.append(
                (pid, "Shape or affine mismatch")
            )

    except Exception as e:
        issues.append((pid, str(e)))

print("Total patients:", len(patients))
print("Valid patients:", len(valid))
print("Patients with issues:", len(issues))

for issue in issues:
    print(issue)

print("\nValidation completed.")
```
