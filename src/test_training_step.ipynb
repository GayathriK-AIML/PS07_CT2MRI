```python

import torch
import torch.nn as nn
from resunet import ResidualUNet

# 1. Select GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Initialize model
model = ResidualUNet(
    in_channels=3,
    out_channels=1,
    base_channels=16
).to(device)

# 3. Training mode
model.train()

# 4. Optimizer and loss
criterion = nn.L1Loss()
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4
)

# 5. Create dummy training batch
ct = torch.rand(1, 3, 192, 224).to(device)
target_mri = torch.rand(1, 1, 192, 224).to(device)

# 6. Forward pass
optimizer.zero_grad()

prediction = model(ct)

# 7. Calculate loss
loss = criterion(prediction, target_mri)

# 8. Backpropagation
loss.backward()

# 9. Update weights
optimizer.step()

print("Training step successful!")
print("Loss:", loss.item())
print("Prediction shape:", prediction.shape)
print("GPU memory allocated:",
      round(torch.cuda.memory_allocated() / 1024**2, 2), "MB")
```
