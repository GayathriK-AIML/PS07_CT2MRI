```python
import nibabel as nib
import numpy as np

BASE = Path(r"D:\PS07_CT2MRI\dataset\volumes")

patient_ids = ["P1046", "P2622", "P2658"]

for pid in patient_ids:

    folder = BASE / pid

    ct = nib.load(str(folder / "ct.nii.gz")).get_fdata()
    mr = nib.load(str(folder / "mr.nii.gz")).get_fdata()
    mask = nib.load(str(folder / "mask.nii.gz")).get_fdata()

    ct_norm = normalise_ct(ct)
    mr_norm = normalise_mri(mr, mask)

    print("\nPatient:", pid)

    print("CT normalized range:",
          ct_norm.min(), "to", ct_norm.max())

    print("MRI normalized range:",
          mr_norm.min(), "to", mr_norm.max())

    print("MRI 99.5 percentile inside mask:",
          np.percentile(mr[mask > 0], 99.5))
```
