```python

import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Patient folder
patient_dir = Path(
    r"D:\PS07_CT2MRI\dataset\volumes\P1046"
)

# Load volumes
ct = nib.load(str(patient_dir / "ct.nii.gz")).get_fdata()
mr = nib.load(str(patient_dir / "mr.nii.gz")).get_fdata()
mask = nib.load(str(patient_dir / "mask.nii.gz")).get_fdata()

# Select middle slice along the third axis
slice_idx = ct.shape[2] // 2

ct_slice = ct[:, :, slice_idx]
mr_slice = mr[:, :, slice_idx]
mask_slice = mask[:, :, slice_idx]

# Display images
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(ct_slice.T, cmap="gray", origin="lower",
               vmin=-1000, vmax=2000)
axes[0].set_title("CT Scan")

axes[1].imshow(mr_slice.T, cmap="gray", origin="lower")
axes[1].set_title("MRI Scan")

axes[2].imshow(mask_slice.T, cmap="gray", origin="lower")
axes[2].set_title("Mask")

for ax in axes:
    ax.axis("off")

plt.suptitle(f"Patient P1046 - Slice {slice_idx}")
plt.tight_layout()
plt.show()
```

```python

from pathlib import Path

import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path(r"D:\PS07_CT2MRI")
VOLUMES = ROOT / "dataset" / "volumes"

pid = "P1046"

folder = VOLUMES / pid

ct = np.asarray(
    nib.load(str(folder / "ct.nii.gz")).dataobj
)

mask = np.asarray(
    nib.load(str(folder / "mask.nii.gz")).dataobj
)

# Select representative slices
slices = [20, 50, 80, 110, 140, 160]

fig, axes = plt.subplots(2, 3, figsize=(13, 8))

for ax, z in zip(axes.flat, slices):

    ax.imshow(ct[:, :, z].T, cmap="gray", origin="lower")

    # Overlay mask boundary
    ax.contour(
        mask[:, :, z].T,
        levels=[0.5],
        colors=["red"],
        linewidths=0.7,
        origin="lower"
    )

    ax.set_title(f"Patient {pid} - Slice {z}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
